```python
import pandas as pd
import numpy as np
```

```python
# Load the first dataframe which is the Books.csv file

books = pd.read_csv('../data/Books.csv',on_bad_lines='skip',encoding='latin-1')
```

```python
books.head()
```

```python
books.columns
```

```python
books.iloc[3]['Image-URL-L']
```

```python
books.info()
```

```python
books = books[['ISBN', 'Book-Title', 'Book-Author', 'Image-URL-L', 'Publisher', 'Year-Of-Publication']]
```

```python
books.head()
```

```python
books.rename(columns={'Book-Title': 'Title', 
                    'Book-Author': 'Author',
                    'Image-URL-L': 'Image_URL',
                    'Year-Of-Publication': 'Year',
                    'Publisher': 'Publisher'
                      }, inplace=True)
```

```python
books.head()
```

```python
## Now load the second dataframe which is the Users.csv file

users = pd.read_csv('../data/Users.csv',on_bad_lines='skip',encoding='latin-1')
```

```python
users.head()
```

```python
users.rename(columns={'User-ID': 'User_ID',
                    'Location': 'Location',
                    'Age': 'Age'
                      }, inplace=True)
```

```python
users.head()
```

```python
# Load the third dataframe which is the Ratings.csv file

ratings = pd.read_csv('../data/Ratings.csv',on_bad_lines='skip',encoding='latin-1')
```

```python
ratings.head()
```

```python
ratings.rename(columns={'User-ID': 'User_ID',
                    'ISBN': 'ISBN',
                    'Book-Rating': 'Rating'
                      }, inplace=True)
```

```python
ratings.head()
```

```python
ratings['User_ID'].value_counts
```

```python
# Users who had atleat given ratings to more than 200 books

x = ratings['User_ID'].value_counts() > 200 
```

```python
x
```

```python
x[x].shape 
```

```python
index = x[x].index
```

```python
index
```

```python
ratings = ratings[ratings['User_ID'].isin(index)]
```

```python
ratings.head()
```

```python
# Now merge the ratings with the books 

ratings_with_books = ratings.merge(books, on='ISBN')
```

```python
ratings_with_books.head()
```

```python
number_of_ratings = ratings_with_books.groupby('Title')['Rating'].count().reset_index()
```

```python
number_of_ratings.head()
```

```python
number_of_ratings.rename(columns={'Rating': 'Number_of_Ratings'}, inplace=True)
```

```python
number_of_ratings.head()
```

```python
final_dataset = ratings_with_books.merge(number_of_ratings, on='Title')
```

```python
final_dataset.head()
```

```python
# Lets find out those books which have atleast 50 ratings

final_dataset = final_dataset[final_dataset['Number_of_Ratings'] >= 50]
```

```python
final_dataset.head()
```

```python
# Drop the duplicates

final_dataset = final_dataset.drop_duplicates(subset=['User_ID', 'Title'], keep='first')
```

```python
# Create a pivot table with the final dataset

pivot_table = final_dataset.pivot_table(columns='User_ID', index='Title', values='Rating')
```

```python
pivot_table
```

```python
pivot_table.shape
```

```python
pivot_table.fillna(0, inplace=True)
```

```python
from scipy.sparse import csr_matrix
```

```python
book_sparse_matrix = csr_matrix(pivot_table)
```

```python
book_sparse_matrix
```

```python
type(book_sparse_matrix)
```

```python
from sklearn.neighbors import NearestNeighbors
model = NearestNeighbors(algorithm = 'brute')
```

```python
model.fit(book_sparse_matrix)
```

```python
pivot_table.iloc[23]
```

```python
distances, suggestions = model.kneighbors(pivot_table.iloc[23].values.reshape(1,-1), n_neighbors=6)
```

```python
distances
```

```python
for i in range(len(suggestions)):
    print(pivot_table.index[suggestions[i]])
```

```python
def recommend_books(book_name):

    book_id = np.where(pivot_table.index == book_name)[0][0]

    distances, suggestions = model.kneighbors(
        pivot_table.iloc[book_id].values.reshape(1, -1),
        n_neighbors=6
    )

    print(f"You searched: {book_name}\n")
    print("Suggested books:\n")

    for i in suggestions[0]:
        print(pivot_table.index[i])
```

```python
books_name = "The Lovely Bones: A Novel"
recommend_books(books_name)
```
